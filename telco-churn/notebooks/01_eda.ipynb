## Temuan Utama

1. **Kelas tidak seimbang**: Churn rate 26.5% (1869 dari 7043 pelanggan). Akurasi bukan metrik yang tepat.

2. **Tenure sangat berpengaruh**: Pelanggan yang churn punya tenure rata-rata 18 bulan, pelanggan setia 38 bulan. Pelanggan baru berisiko tinggi.

3. **Month-to-month contract dominan churn**: Churn rate 42.7% vs one year 11.3% vs two year 2.8%. Kontrak jangka panjang adalah proteksi kuat.

4. **Fiber optic paradox**: Pelanggan fiber optic punya churn rate 41.9%, DSL hanya 18.9%. Kemungkinan harga lebih mahal atau ekspektasi layanan lebih tinggi.

5. **Tech support menurunkan churn**: Pelanggan tanpa tech support churn 41.6%, dengan tech support 15.2%. Add-on support bernilai tinggi.

6. **MonthlyCharges berkorelasi positif dengan churn**: Pelanggan yang churn bayar lebih mahal per bulan, tapi TotalCharges lebih rendah (tenure pendek).

7. **Payment method electronic check berisiko**: Perlu cek lebih lanjut pada modeling.
